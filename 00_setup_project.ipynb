```python
import requests

urls = {
    "NLR": (
        "https://oedi-data-lake.s3.amazonaws.com/"
        "pvdaq/2023-solar-data-prize/2107_OEDI/"
        "metadata/2107_system_metadata.json"
    ),
    "OpenMeteo": "https://api.open-meteo.com/v1/forecast?latitude=39&longitude=-122",
    "CAISO": "https://www.caiso.com"
}

for name, url in urls.items():
    try:
        response = requests.get(url, timeout=20)
        print(name, response.status_code, len(response.content))
    except Exception as exc:
        print(name, "ERROR", type(exc).__name__, str(exc)[:200])
```
